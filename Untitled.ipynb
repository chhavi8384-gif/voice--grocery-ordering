```python
import pandas as pd
from rapidfuzz import process, fuzz

catalog = pd.DataFrame([
    {"item": "rice",  "shop": "Shop A", "price_per_kg": 60},
    {"item": "dal",   "shop": "Shop A", "price_per_kg": 120},
    {"item": "sugar", "shop": "Shop A", "price_per_kg": 45},
    {"item": "rice",  "shop": "Shop B", "price_per_kg": 65},
    {"item": "dal",   "shop": "Shop B", "price_per_kg": 115},
    {"item": "sugar", "shop": "Shop B", "price_per_kg": 48},
])

# Common Hindi words/transcriptions mapped to catalog names
aliases = {
    "chawal": "rice",
    "chaawal": "rice",
    "चावल": "rice",
    "daal": "dal",
    "दाल": "dal",
    "chini": "sugar",
    "चीनी": "sugar",
}

def find_item(spoken_item):
    query = aliases.get(spoken_item.strip().lower(), spoken_item.strip().lower())
    choices = catalog["item"].unique().tolist()
    match = process.extractOne(query, choices, scorer=fuzz.ratio)

    if match and match[1] >= 65:
        return match[0]
    return None

print(find_item("chawal"))  # should print: rice
print(find_item("दाल"))     # should print: dal
```

```python
import json

def extract_items(order_text):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "system",
                "content": (
                    "Extract grocery items and quantities from the user's order. "
                    "Return only valid JSON in this exact shape: "
                    '{"items":[{"item":"chawal","qty":"0.5 kg"}]}. '
                    "Use null for qty if the user did not specify an amount. "
                    "Keep item names as spoken."
                ),
            },
            {"role": "user", "content": order_text},
        ],
        response_format={"type": "json_object"},
    )

    return json.loads(response.choices[0].message.content)

result = extract_items("Mujhe aadha kilo chawal aur ek kilo dal chahiye")
print(result)
```

```python
import re

def quantity_in_kg(qty):
    if not qty:
        return None

    text = str(qty).lower().strip()
    match = re.search(r"(\d+(?:\.\d+)?)", text)

    if not match:
        return None

    amount = float(match.group(1))

    if any(unit in text for unit in ["gram", "grams", " g"]):
        return amount / 1000

    if any(unit in text for unit in ["kg", "kilo"]):
        return amount

    return None  # Unknown unit; don't guess

def build_quote(extracted):
    quote = []

    for entry in extracted["items"]:
        matched_item = find_item(entry["item"])
        if not matched_item:
            quote.append({
                "requested_item": entry["item"],
                "status": "not found in catalog",
            })
            continue

        # Select the shop with the lowest price for this item
        options = catalog[catalog["item"] == matched_item]
        best = options.loc[options["price_per_kg"].idxmin()]
        kg = quantity_in_kg(entry.get("qty"))

        quote.append({
            "item": matched_item,
            "quantity": entry.get("qty"),
            "shop": best["shop"],
            "price_per_kg": best["price_per_kg"],
            "subtotal": round(kg * best["price_per_kg"], 2) if kg is not None else None,
        })

    return quote

quote = build_quote(result)
print(quote)
```

```python
from IPython.display import Javascript, display
from google.colab import output
import base64

display(Javascript("""
async function recordAudio(seconds) {
  const stream = await navigator.mediaDevices.getUserMedia({audio: true});
  const recorder = new MediaRecorder(stream);
  const chunks = [];

  recorder.ondataavailable = event => chunks.push(event.data);
  recorder.start();

  await new Promise(resolve => setTimeout(resolve, seconds * 1000));
  recorder.stop();
  await new Promise(resolve => recorder.onstop = resolve);
  stream.getTracks().forEach(track => track.stop());

  const blob = new Blob(chunks, {type: "audio/webm"});
  const reader = new FileReader();
  reader.readAsDataURL(blob);
  return await new Promise(resolve => reader.onloadend = () => resolve(reader.result));
}
"""))

audio_data = output.eval_js("recordAudio(8)")
audio_bytes = base64.b64decode(audio_data.split(",")[1])

with open("voice.webm", "wb") as f:
    f.write(audio_bytes)

audio_path = "voice.webm"
print("Recording saved:", audio_path)
```

```python
import json
import os

memory_file = "/content/last_order.json"

def save_last_order(order):
    with open(memory_file, "w", encoding="utf-8") as f:
        json.dump(
            order,
            f,
            ensure_ascii=False,
            default=lambda value: value.item() if hasattr(value, "item") else str(value),
        )

def load_last_order():
    if not os.path.exists(memory_file):
        return None

    with open(memory_file, "r", encoding="utf-8") as f:
        return json.load(f)

# Save the order you just tested
save_last_order(quote)

print("Saved last order:")
print(load_last_order())
```

```python
def is_reorder_request(text):
    text = text.lower()
    phrases = [
        "reorder as usual",
        "same as last time",
        "repeat my last order",
        "usual order",
        "wahi order",
        "pichla order",
        "pehle wala order",
    ]
    return any(phrase in text for phrase in phrases)

def get_quote_for_text(text):
    if is_reorder_request(text):
        previous_order = load_last_order()
        if previous_order is not None:
            return previous_order
        raise ValueError("There is no saved previous order yet.")

    extracted = extract_items(text)
    new_quote = build_quote(extracted)
    save_last_order(new_quote)
    return new_quote

test_quote = get_quote_for_text("reorder as usual")
print(test_quote)
```

```python
import subprocess
from gtts import gTTS
from IPython.display import Audio, display

def process_audio_order(audio_path, show_audio=True):
    # Voice → text
    segments, info = whisper.transcribe(audio_path, language="hi")
    spoken_text = " ".join(segment.text for segment in segments).strip()

    if not spoken_text:
        raise ValueError("No speech detected. Please check the recording.")

    print("You said:", spoken_text)

    # Text → catalog quote (or load saved order)
    order = get_quote_for_text(spoken_text)

    # Quote → Hindi confirmation
    item_names_hi = {"rice": "चावल", "dal": "दाल", "sugar": "चीनी"}
    lines = []
    total = 0

    for item in order:
        if item.get("subtotal") is None:
            lines.append(f'{item.get("requested_item", item["item"])} की कीमत नहीं मिली')
            continue

        name = item_names_hi.get(item["item"], item["item"])
        subtotal = float(item["subtotal"])
        total += subtotal
        lines.append(f'{item["quantity"]} {name}, {subtotal:.0f} रुपये')

    confirmation = (
        "आपका ऑर्डर है: " + "। ".join(lines) +
        f"। कुल अनुमानित कीमत {total:.0f} रुपये है।"
    )

    print("Confirmation:", confirmation)

    # Hindi text → audio; convert to WAV for reliable Colab playback
    gTTS(confirmation, lang="hi").save("confirmation.mp3")
    subprocess.run(
        ["ffmpeg", "-y", "-i", "confirmation.mp3", "confirmation.wav", "-loglevel", "error"],
        check=True,
    )
    display(Audio(filename="confirmation.wav"))

    return order

order = process_audio_order("/content/audio.mp4")
```

```python
import os
import ipywidgets as widgets
from IPython.display import display, clear_output

audio_upload = widgets.FileUpload(
    accept=".mp3,.wav,.m4a,.mp4,.webm",
    multiple=False,
    description="Choose audio",
)

run_button = widgets.Button(
    description="Place order",
    button_style="success",
)

result_area = widgets.Output()

def run_demo(_):
    with result_area:
        clear_output()
        uploaded = audio_upload.value

        if not uploaded:
            print("Choose an audio file first.")
            return

        # Support both common Colab widget upload formats
        if isinstance(uploaded, dict):
            filename, details = next(iter(uploaded.items()))
        else:
            details = uploaded[0]
            filename = details["name"]

        audio_bytes = bytes(details["content"])
        safe_name = os.path.basename(filename)
        audio_path = f"/content/{safe_name}"

        with open(audio_path, "wb") as f:
            f.write(audio_bytes)

        print("Processing:", safe_name)
        process_audio_order(audio_path)

run_button.on_click(run_demo)
display(widgets.VBox([audio_upload, run_button, result_area]))
```
